## Bibliotecas

```python
import pandas as pd
```

## URL Usada

```python

url = "https://www.basketball-reference.com/contracts/players.html"
```

## Scrapping do DataFrame

```python
df = pd.read_html(url, header=[0, 1])[0]
```

## Tratamento com as colunas

```python
new_cols = []
for top, bot in df.columns:
    top = str(top)
    bot = str(bot)

    if top.startswith("Unnamed"):          # usa apenas o nível de baixo
        name = bot
    elif bot.startswith("Unnamed"):        # usa apenas o nível de cima
        name = top
    elif top == "Salary":                  # em 'Salary', mantenha só o ano
        name = bot
    else:                                  # demais casos, concatene
        name = f"{top} {bot}"

    new_cols.append(name.strip())

df.columns = new_cols
```
